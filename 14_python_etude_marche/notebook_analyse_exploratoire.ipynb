```python
#Tableau temporaire
df_pivot_pib = df_fao.pivot_table(index='Pays',columns='Année',values='PIB/hab ($ PPA internationaux constants de 2021)')

#Calcul du TCAM entre 2017 et 2022 : ((V_finale / V_initiale)^(1/n) - 1) * 100
n = 7
df_pivot_pib['TCAM PIB 2017-2022'] = ((df_pivot_pib[2022] / df_pivot_pib[2017])**(1/n) - 1) * 100

#Ajout à df_fao
df_fao = pd.merge(df_fao,df_pivot_pib[['TCAM PIB 2017-2022']], on='Pays', how='left')
df_fao
```

## 4.2 - FAO : Taux de croissance annuel moyen de la population entre 2017 et 2022

```python
#Tableau temporaire
df_pivot_pop = df_fao.pivot_table(index='Pays',columns='Année',values='Pop totale (pour 1000)')

#Calcul du TCAM entre 2017 et 2022 : ((V_finale / V_initiale)^(1/n) - 1) * 100
n = 7
df_pivot_pop['TCAM Population 2017-2022'] = ((df_pivot_pop[2022] / df_pivot_pop[2017])**(1/n) - 1) * 100

#Ajout à df_fao
df_fao = pd.merge(df_fao,df_pivot_pop[['TCAM Population 2017-2022']], on='Pays', how='left')
df_fao
```

## 4.3 - FAO : Taux d'urbanisation annuel

```python
df_fao['Taux urbanisation annuel'] = (df_fao['Pop urbaine (pour 1000)'] / df_fao['Pop totale (pour 1000)']) * 100
df_fao = df_fao.drop(columns=['Pop urbaine (pour 1000)','Pop rurale (pour 1000)'])
df_fao
```

## 4.4 - FAO : Taux d'autosuffisance

```python
df_fao['Taux autosuffisance'] = (df_fao['Production viande volaille (1000 t)']/df_fao['Dispo intérieure viande volaille (1000 t)']) * 100
df_fao = df_fao.drop(columns=['Production viande volaille (1000 t)','Dispo intérieure viande volaille (1000 t)'])
df_fao
```

## 4.5 - CEPII : Score langue

```python
#Création d'une fonction personnalisée à appliquer à chaque ligne
def score_langue(row):
#Création d'une liste des langues pour par ligne
    langues = [str(row['langoff_1']).lower(),
               str(row['langoff_2']).lower(),
               str(row['langoff_3']).lower()]

    # Condition 1 : français présent
    if any("français" in l or "french" in l for l in langues):
        return 2
    # Condition 2 : anglais présent (et donc pas français d'après l'ordre du IF)
    elif any("anglais" in l or "english" in l for l in langues):
        return 1
    # Condition 3 : Reste
    else:
        return 0

# Application de la fonction à df_fao
df_cepii['Score Langue'] = df_cepii.apply(score_langue, axis=1)
df_cepii
```

## 6.2 - Identification des outliers

```python
#Identification
noms_pays = df_f3['Pays'].values
df_num = df_f3.select_dtypes(include=[np.number])

mean = df_num.mean()
std = df_num.std()

z_scores = (df_num - mean) / std

anomalies = []

for i in range(len(z_scores)):
    row = z_scores.iloc[i]
    for column, value in row.items():
        if abs(value) > 5:
            anomalies.append({
                'Pays': noms_pays[i],
                'Variable': column,
                'Valeur Brute': df_num.iloc[i][column],
                'Z-score': round(value, 2)
            })

df_anomalies_detail = pd.DataFrame(anomalies)

df_anomalies_detail.sort_values(by='Z-score', ascending=False)
```
